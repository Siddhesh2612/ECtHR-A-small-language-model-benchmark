```python
import matplotlib.pyplot as plt
import numpy as np

# Models and their scores
models = [
    'GPT-2\n(1024 tokens)',
    'GPT-Neo 125M\n(2048 tokens)',
    'GPT-Neo 1.3B\n(1024 tokens)',
    'Phi-2\n(1024 tokens)',
    'TF-IDF',
    'Legal-BERT'
]

metrics = ['Micro F1', 'Macro F1', 'Weighted F1']

# Updated scores
scores = {
    'GPT-2\n(1024 tokens)': [0.670, 0.579, 0.653],
    'GPT-Neo 125M\n(2048 tokens)': [0.675, 0.548, 0.660],
    'GPT-Neo 1.3B\n(1024 tokens)': [0.6525, 0.527, 0.630],
    'Phi-2\n(1024 tokens)': [0.635, 0.501, 0.612],
    'TF-IDF': [0.626, 0.489, 0.000],
    'Legal-BERT': [0.700, 0.640, 0.000]
}

# Number of variables we're plotting
num_vars = len(metrics)

# Compute angle each axis is at
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

# Initialize the radar plot
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Define colors for each model
colors = ['blue', 'green', 'red', 'purple', 'orange', 'cyan']

# Plot each model
for idx, model in enumerate(models):
    values = scores[model]
    values += values[:1]  # Close the loop for the model
    ax.plot(angles, values, color=colors[idx], label=model, linewidth=2)
    ax.fill(angles, values, color=colors[idx], alpha=0.25)

# Add the attribute labels to our axes
ax.set_thetagrids(np.degrees(angles[:-1]), metrics)

# Set the range for the radial axis
ax.set_ylim(0, 1)

# Add a legend
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

# Add a title
#plt.title('Model Performance Comparison')

# Show the plot
plt.tight_layout()
plt.show()
```
